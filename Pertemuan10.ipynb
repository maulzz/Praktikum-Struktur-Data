```python
class SimpleHashTable:
  def __init__(self, size):
    self.size = size
    self.table = [None] * size

  def hash_function(self,key):
    return hash(key) % self.size

  def insert (self, key, value):
    index = self.hash_function(key)
    self.table[index] = value

  def search(self,key):
    index = self.hash_function(key)
    return self.table[index]

ht = SimpleHashTable(10)
ht.insert("Apel Hijau Banyumas", 100)
ht.insert("Pisang Mas Amba", 200)

print("Nilai untuk Apel Hijau Banyumas: ", ht.search("Apel Hijau Banyumas"))
print("Nilai untuk Pisang Mas Amba: ", ht.search("Pisang Mas Amba"))
```

```python
class ChainingHashTable:
  def __init__(self, size):
    self.size = size
    self.table = [[] for _ in range(size)]

  def hash_function(self, key):
    return hash(key) % self.size

  def insert(self, key, value):
    index = self.hash_function(key)
    for i, (k,v) in enumerate(self.table[index]):
      if k == key:
        self.table[index] [i] = (key, value)
        return
    self.table[index].append((key, value))

  def search(self, key):
    index = self.hash_function(key)
    for k, v in self.table[index]:
      if k == key:
        return v
      return None

ht = ChainingHashTable(5)
ht.insert("Apel Hijau Banyumas", 100)
ht.insert("Pisang Mas Amba", 200)
ht.insert("Puding Pak Hambali", 400)

print("Nilai untuk Apel Hijau Banyumas: ", ht.search("Apel Hijau Banyumas"))
print("Nilai untuk Pisang Mas Amba: ", ht.search("Pisang Mas Amba"))
```

```python
class LinearProbingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size

    # Fungsi hash sederhana
    def hash_function(self, key):
        return ord(key[0]) % self.size

    # Insert data dengan linear probing
    def insert(self, key, value):
        index = self.hash_function(key)

        # Jika slot penuh, cari slot berikutnya
        while self.table[index] is not None:
            index = (index + 1) % self.size

        self.table[index] = (key, value)

    # Menampilkan isi tabel
    def display(self):
        print("Isi Hash Table:")
        for i, data in enumerate(self.table):
            print(f"Index {i}: {data}")


hash_table = LinearProbingHashTable(5)

hash_table.insert("A", 10)
hash_table.insert("B", 20)
hash_table.insert("C", 30)
hash_table.insert("D", 40)
hash_table.insert("E", 50)

hash_table.display()
```

```python
class ChainingHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    # Fungsi hash sederhana
    def hash_function(self, key):
        return len(key) % self.size

    # Insert data
    def insert(self, key, value):
        index = self.hash_function(key)

        # Tambahkan data ke bucket
        self.table[index].append((key, value))

    # Menampilkan isi tabel
    def display(self):
        print("Isi Hash Table:")
        for i, bucket in enumerate(self.table):
            print(f"Index {i}: {bucket}")


hash_table = ChainingHashTable(3)

hash_table.insert("apel", 100)
hash_table.insert("melon", 200)
hash_table.insert("lemon", 300)
hash_table.insert("pisang", 400)

hash_table.display()
```
