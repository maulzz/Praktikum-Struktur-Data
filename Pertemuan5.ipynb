```python
def LinearSearch(arr,n,x):
  for i in range(0,n):
    if(arr[i] == x):
      return i
  return -1

if __name__ == '__main__':
  arr = [4,2,22,44,13,67,11,34,7]
  X = 13
  N = len(arr)

  hasil = LinearSearch(arr,N,X)
  if(hasil == -1):
    print("elemen yang dicari tidak ada dalam array")
  else:
    print('elemen yang dicari berada di indeks ke-',hasil)
```

```python
def BinarySeacrh(arr, l,r,x):

  while l <= r:
    mid = l +(r-l) // 2
    if arr[mid] == x:
      return mid
    elif arr[mid] < x:
      l = mid + 1
    else:
      r = mid-1
  return -1


if __name__ == '__main__':
  arr =[2,5,7,12,16,23,28,30,40,53,67,71,82]
  X = 28

  hasil = BinarySeacrh(arr, 0, len(arr)-1,X)
  if hasil != -1:
    print("elemen yang dicari berada di indeks ke-",hasil)
  else:
    print("elemn yang dicari tidak ada didalam array")
```

```python
def BinarySeacrh(arr, l,r,x):

  if l <= r:
    mid = l +(r-1) // 2
    if arr[mid] == x:
      return mid
    elif arr[mid] > x:
      return BinarySeacrh(arr, 1, mid-1, x)
    else:
      return BinarySeacrh(arr,mid +1,r,x)
  else:
    return -1


if __name__ == '__main__':
  arr =[2,5,7,12,16,23,28,30,40,53,67,71,82]
  X = 67

  hasil = BinarySeacrh(arr, 0, len(arr)-1,X)
  if hasil != -1:
    print("elemen yang dicari berada di indeks ke-",hasil)
  else:
    print("elemn yang dicari tidak ada didalam array")
```

```python
import math

def JumpSeacrh (arr, x, n):
  step = math.sqrt(n)

  prev = 0
  while arr[int(min(step,n)-1)] < x:
    prev = step
    step += math.sqrt(n)
    if prev >= n:
      return -1

  while arr[int(prev)] < x:
    prev += 1

    if prev == min(step,n):
      return -1

  if arr[int(prev)] == x:
    return prev
  return -1


arr = [0,1,1,2,3,3,5,5,33,75,80,91,145,235]
x = 80
n = len(arr)

index = JumpSeacrh(arr,x,n)
print(f'angka {x} ,berada diindex', "%.0f"%index)
```

```python
class pair:
  def __init__(self):
    self.min = 0
    self.max = 0


def getMinMax(arr: list, n: int) -> pair:
  minmax = pair()

  if n == 1:
    minmax.max = arr[0]
    minmax.min = arr[1]
    return minmax

  if arr[0] > arr[1]:
    minmax.max = arr[0]
    minmax.min = arr[1]
  else:
    minmax.max = arr[0]
    minmax.min = arr[1]

  for i in range(2,n):
    if arr[i] > minmax.max:
      minmax.max = arr[i]
    elif arr[i] < minmax.min:
      minmax.min = arr[i]
  return minmax


if __name__ == '__main__':
  arr = [1000, 11,100,445,23,1,30000]
  arr_size = len(arr)
  minmax = getMinMax(arr,arr_size)
  print(f'minimun element is {minmax.min}')
  print(f'minimun element is {minmax.max}')


```
