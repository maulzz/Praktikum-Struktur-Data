```python
graph = {
    'A' : ['B', 'C'],
    'B' : ['A', 'D', 'E'],
    'C' : ['A', 'F'],
    'D' : ['B',],
    'E' : ['B', 'F'],
    'F' : ['C', 'E'],
}

def print_graph(graf):
  for node in graf:
    print(f'{node} -> {graf[node]}')

print_graph(graph)
```

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque([start])

  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end='')
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

  bfs(graph, 'A')
```

```python
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set()
  if start not in visited:
    print(start, end='')
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

dfs(graph, 'A')
```

```python
def count_neighbors(graph):
    result = ()
    for node in graph:
        # hitung jumlah tetangga
        result += (len(graph[node]),)
    return result

print(count_neighbors(graph))
```

```python
from collections import deque

class Graph:
    def __init__(self):
        self.graph = {}

    # Tambah simpul
    def add_vertex(self, vertex):
        if vertex not in self.graph:
            self.graph[vertex] = []

    # Tambah sisi (tak berarah)
    def add_edge(self, u, v):
        if u not in self.graph:
            self.add_vertex(u)
        if v not in self.graph:
            self.add_vertex(v)

        self.graph[u].append(v)
        self.graph[v].append(u)

    # Cetak graph
    def print_graph(self):
        print("Graph:")
        for node in self.graph:
            print(f"{node} -> {self.graph[node]}")

    # BFS (mengembalikan urutan kunjungan)
    def bfs(self, start):
        visited = set()
        queue = deque([start])
        order = []

        visited.add(start)

        while queue:
            node = queue.popleft()
            order.append(node)

            for neighbor in self.graph[node]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)

        return order

    # DFS
    def dfs(self, start, visited=None, order=None):
        if visited is None:
            visited = set()
        if order is None:
            order = []

        visited.add(start)
        order.append(start)

        for neighbor in self.graph[start]:
            if neighbor not in visited:
                self.dfs(neighbor, visited, order)

        return order

    # Cari satu jalur dari start ke end (DFS)
    def find_path(self, start, end, visited=None, path=None):
        if visited is None:
            visited = set()
        if path is None:
            path = []

        visited.add(start)
        path.append(start)

        if start == end:
            return path

        for neighbor in self.graph[start]:
            if neighbor not in visited:
                result = self.find_path(neighbor, end, visited, path.copy())
                if result:
                    return result

        return None

    # Cek apakah semua simpul terhubung
    def is_connected(self):
        nodes = list(self.graph.keys())
        if not nodes:
            return True

        visited_nodes = self.dfs(nodes[0])
        return len(visited_nodes) == len(self.graph)

g = Graph()

# Menambahkan 7 sisi
g.add_edge('A', 'B')
g.add_edge('A', 'C')
g.add_edge('B', 'C')
g.add_edge('B', 'D')
g.add_edge('C', 'D')
g.add_edge('C', 'E')
g.add_edge('D', 'E')

# Cetak graph
g.print_graph()

# BFS dan DFS
print("\nBFS dari A:", g.bfs('A'))
print("DFS dari A:", g.dfs('A'))

# Cari jalur
print("\nPath dari A ke E:", g.find_path('A', 'E'))

# Cek konektivitas
print("\nApakah graph terhubung?", g.is_connected())
```
