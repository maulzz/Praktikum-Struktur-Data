```python
queue = []

queue.append('1')
queue.append('2')
queue.append('3')
queue.append('4')
queue.append('5')
queue.append('6')

print("Isi queue")
print(queue)

print("\nitem yang dihapus dari queue")
print(queue.pop(0))
print(queue.pop(0))
print(queue.pop(0))

print("\nIsi queue")
print(queue)
```

```python
from collections import deque

queue = deque()

queue.append('1')
queue.append('2')
queue.append('3')
queue.append('4')
queue.append('5')
queue.append('6')

print("Isi queue")
print(queue)

print("\nitem yang dihapus dari queue")
print(queue.popleft())
print(queue.popleft())
print(queue.popleft())

print("\nIsi queue")
print(queue)
```

```python
from queue import Queue

qu = Queue(maxsize=6)

print(qu.qsize())

qu.put('1')
qu.put('2')
qu.put('3')
qu.put('4')
qu.put('5')
qu.put('6')

print('full:', qu.full())
print('Size:', qu.qsize())

print("\nitem yang dihapus dari queue")
print(qu.get())
print(qu.get())
print(qu.get())

print('\nEmpty:', qu.empty())
print('full:', qu.full())
```

```python
from typing import Self
class Queue:
  def __init__(self):
    self.queue = list()

  def addtoqu(self,dataval):
    if dataval not in self.queue:
      self.queue.insert(0, dataval)
      return True
    return False

  def removeFromq(self):
    if len(self.queue)>0:
      return self.queue.pop()
    return ("Tidak ada element didalam queue!")

  def size(self):
    return len(self.queue)

TheQueue = Queue()
print(TheQueue.addtoqu('1'))
print(TheQueue.addtoqu('2'))
print(TheQueue.addtoqu('3'))
print(TheQueue.addtoqu('4'))
print(TheQueue.size())
print(TheQueue.removeFromq())
print(TheQueue.removeFromq())
print(TheQueue.size())
```
