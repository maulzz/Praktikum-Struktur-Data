```python
def bubbleSort(arr):
  n = len(arr)
  for i in range (n-1):
    for j in range(n - i - 1):
      if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j+1], arr[j]
  return arr

data = [4,7,22,55,43,67,84,19,27]
sortedData = bubbleSort(data.copy())
print(f'data sebelum di sorting: {data}')
print(f'data setelah bubble sort: {sortedData}')
```

```python
def selectionSort(arr):
  n = len(arr)
  for i in range (n-1):
    minIndex = i
    for j in range(n + 1, n):
      if arr[j] < arr[minIndex]:
        minIndex = j
    arr[i], arr[minIndex] = arr[minIndex], arr[i]
  return arr

data = [4,7,22,55,43,67,84,19,27]
sortedData = selectionSort(data.copy())
print(f'data sebelum di sorting: {data}')
print(f'data setelah Selection Sort: {sortedData}')
```

```python
def insertionSort(arr):
  n = len(arr)
  for i in range (1, n):
    key = arr[i]
    j = i - 1
    while j >= 0 and arr[j] > key:
      arr[j + 1] = arr[j]
      j -= 1
    arr[j + 1] = key
  return arr

data = [4,7,22,55,43,67,84,19,27]
sortedData = insertionSort(data.copy())
print(f'data sebelum di sorting: {data}')
print(f'data setelah Insertion Sort: {sortedData}')
```

```python
def mergeSort(arr):
  if len(arr) > 1:
    mid = len(arr) // 2
    leftHalf = arr[:mid]
    rightHalf = arr[mid:]

    mergeSort(leftHalf)
    mergeSort(rightHalf)

    i = j = k = 0

    while i < len(leftHalf) and j < len(rightHalf):
      if leftHalf[i] < rightHalf[j]:
        arr[k] = leftHalf[i]
        i += 1
      else:
        arr[k] = rightHalf[j]
        j += 1
        k += 1

    while i < len(leftHalf):
      arr[k] = leftHalf[i]
      i += 1
      k += 1

    while i < len(rightHalf):
      arr[k] = rightHalf[j]
      j += 1
      k += 1
  return arr

data = [4,7,22,55,43,67,84,19,27]
sortedData = mergeSort(data.copy())
print(f'data sebelum di sorting: {data}')
print(f'data setelah Insertion Sort: {sortedData}')
```

```python
def quickSort(arr):
  if len(arr) <= 1:
    return arr

  pivot = arr[len(arr) // 2]
  left = [x for x in arr if x < pivot]
  middle = [x for x in arr if x == pivot]
  right = [x for x in arr if x > pivot]

  return quickSort(left) + middle + quickSort(right)

data = [4,7,22,55,43,67,84,19,27]
sortedData = quickSort(data.copy())
print(f'data sebelum di sorting: {data}')
print(f'data setelah Insertion Sort: {sortedData}')

```

```python
products = [
    {"nama": "Kaos Kaki MBG", "harga": 5000000, "rating": 2.5},
    {"nama": "Server Komdigi", "harga": 100000000, "rating": 3.7},
    {"nama": "Keyboard Mechanical", "harga": 450000, "rating": 4.6},
    {"nama": "Monitor Oled", "harga": 6100000, "rating": 4.4},
    {"nama": "Headset Gaming", "harga": 500000, "rating": 4.3},
    {"nama": "SSD 512Gb Nvme", "harga": 1100000, "rating": 4.8},
    {"nama": "Flashdisk 64GB", "harga": 90000, "rating": 4.2}
]

def quick_sort(data, key, reverse=False):
    if len(data) <= 1:
        return data
    else:
        pivot = data[0]
        left = []
        right = []
        for item in data[1:]:
            if reverse:
                if item[key] > pivot[key]:
                    left.append(item)
                else:
                    right.append(item)
            else:
                if item[key] < pivot[key]:
                    left.append(item)
                else:
                    right.append(item)
        return quick_sort(left, key, reverse) + [pivot] + quick_sort(right, key, reverse)

sorted_by_price = quick_sort(products, "harga")
sorted_by_rating = quick_sort(products, "rating", reverse=True)

print("=== Produk Berdasarkan Harga (Terendah ke Tertinggi) ===")
for p in sorted_by_price:
    print(f"{p['nama']} - Rp{p['harga']} - Rating: {p['rating']}")

print("\n=== Produk Berdasarkan Rating (Tertinggi ke Terendah) ===")
for p in sorted_by_rating:
    print(f"{p['nama']} - Rp{p['harga']} - Rating: {p['rating']}")
```
